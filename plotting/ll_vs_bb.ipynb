## UMA

```python
import os

DESIRED_ORDER = ["OC20", "ODAC", "OMC", "OMOL", "OMAT"]

HEAD_MAPPING = {
    "OC20": "OC20",
    "ODAC": "ODAC",
    "OMAT": "OMAT",
    "OMC": "OMC",
    "OMOL": "OMOL",
}

base = "../results/reconstruction_errors/ll_vs_bb"
results_gfre_bb = load_results(os.path.join(base, "gfre_uma.json"))
results_lfre_bb = load_results(os.path.join(base, "lfre_uma.json"))

error_matrix_gfre, annot_matrix_gfre = get_2xN_matrix(
    results_gfre_bb, DESIRED_ORDER, HEAD_MAPPING
)
error_matrix_lfre, annot_matrix_lfre = get_2xN_matrix(
    results_lfre_bb, DESIRED_ORDER, HEAD_MAPPING
)
```
