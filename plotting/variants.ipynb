## MACE

```python
import os

DESIRED_ORDER = ["MPTRJ", "OMAT24", "MATPES-PBE", "MATPES-R2SCAN"]

MODEL_MAPPING = {
    "MATPES-PBE": "MATPES-PBE",
    "MATPES-R2SCAN": "MATPES-R2SCAN",
    "OMAT": "OMAT24",
    "MPTRAJ": "MPTRJ",
}

base = "../results/reconstruction_errors/variants/mace"

results_gfre = load_results(os.path.join(base, "gfre_mace-variants.json"))
results_lfre = load_results(os.path.join(base, "lfre_mace-variants.json"))

error_matrix_gfre, annot_matrix_gfre = get_matrices(
    results_gfre, DESIRED_ORDER, MODEL_MAPPING
)
error_matrix_lfre, annot_matrix_lfre = get_matrices(
    results_lfre, DESIRED_ORDER, MODEL_MAPPING
)

plot_bidirectional_heatmaps(
    error_matrix_gfre,
    annot_matrix_gfre,
    error_matrix_lfre,
    annot_matrix_lfre,
    title1="GFRE",
    title2="LFRE",
    cbar_label="GFRE or LFRE",
    rotate_ticks=True,
    save_path="../results/figures/variants/mace_variants.png",
)
```

```python
import os

DESIRED_ORDER = ["SMALL", "MEDIUM", "LARGE"]

MODEL_MAPPING = {
    "SMALL": "SMALL",
    "MEDIUM": "MEDIUM",
    "LARGE": "LARGE",
}

base = "../results/reconstruction_errors/variants/mace/sizes"

results_gfre_mad = load_results(os.path.join(base, "gfre_mace-mp-a0.json"))
results_lfre_mad = load_results(os.path.join(base, "lfre_mace-mp-a0.json"))

results_gfre_organic = load_results(os.path.join(base, "gfre_mace-off23.json"))
results_lfre_organic = load_results(os.path.join(base, "lfre_mace-off23.json"))

error_gfre_mad, annot_gfre_mad = get_matrices(
    results_gfre_mad, DESIRED_ORDER, MODEL_MAPPING
)
error_lfre_mad, annot_lfre_mad = get_matrices(
    results_lfre_mad, DESIRED_ORDER, MODEL_MAPPING
)

error_gfre_org, annot_gfre_org = get_matrices(
    results_gfre_organic, DESIRED_ORDER, MODEL_MAPPING
)
error_lfre_org, annot_lfre_org = get_matrices(
    results_lfre_organic, DESIRED_ORDER, MODEL_MAPPING
)
```

```python
plot_split_diagonal_heatmaps(
    error_gfre_mad,
    error_gfre_org,
    error_lfre_mad,
    error_lfre_org,
    xlabel="target",
    ylabel="source",
    cbar_label="reconstruction error",
    save_path="../results/figures/variants/mace_sizes.png",
)
```
