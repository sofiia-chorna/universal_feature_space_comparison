```python
from utils import plot_bidirectional_heatmaps, get_matrices, load_results
```

## UMLIPs recontruction errors on MAD

```python
import os

base = "../results/reconstruction_errors/umlips"

results_gfre_bb = load_results(os.path.join(base, "gfre_mad.json"))
results_lfre_bb = load_results(os.path.join(base, "lfre_mad.json"))

error_matrix_gfre, annot_matrix_gfre = get_matrices(results_gfre_bb)
error_matrix_lfre, annot_matrix_lfre = get_matrices(results_lfre_bb)

plot_bidirectional_heatmaps(
    error_matrix_gfre, annot_matrix_gfre,
    error_matrix_lfre, annot_matrix_lfre,
    title1="GFRE",
    title2="LFRE",
    cbar_label="GFRE or LFRE",
    save_path="../results/figures/umlips/errors_mad.png",
)
```
